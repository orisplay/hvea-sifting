# HVEA-SIFTING method architecture figure (serif / Times)

Digital-forensics workflow figure. Serif fonts (Times New Roman on Windows) so the
math symbols and text match the Electronics body. Run both code cells top to bottom;
writes **PDF** and **PNG** into the repo's `Images/` folder (run from `~/paperiii-repo`). Dependency: matplotlib (`hvea_env`).

**Symbols:** $I$ = EWF source image · $G_{os}$ = OS-specific HVEA dependency graph ·
$R$ = reasoning trace · $\Pi$ = SIFTING acquisition profile · $P$ = partial image ·
$V$ = validation report.

```python
"""
HVEA-SIFTING method architecture — redesigned.

Changes from v1:
  - No disk/graph clip-art. Inputs are clean labelled tiles with a thin accent bar.
  - Phase-1 reasoner label pushed toward the box (icon removed, text centred).
  - Negative-space audit sits DIRECTLY BELOW the reasoning trace (vertical branch).
  - Bolder, darker text for readability.
  - DF-workflow aesthetic: left accent bars, soft phase bands, consistent hierarchy.

Symbols:
  I     = EWF source image (input)
  G_os  = OS-specific HVEA dependency graph (input)
  R     = reasoning trace (Phase 1 output)
  Pi    = SIFTING acquisition profile (Phase 2 output)
  P     = partial image (Phase 3 output)
  V     = validation report (Phase 4 output)
"""

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Rectangle
import numpy as np

# ---- Output location (EDIT if your figures folder differs) ----
OUT_DIR = "Images"   # repo-relative; run from ~/paperiii-repo
# OUT_DIR = "."

# ---- Palette: three roles, print-safe, DF-workflow feel ----
INPUT_FC, INPUT_BAR = "#EAF2EC", "#3E7D50"   # input tiles (green accent)
PROC_FC,  PROC_BAR  = "#ECEAF8", "#5A4FB0"   # processing (indigo accent)
DATA_FC,  DATA_BAR  = "#F3F2EE", "#7A7870"   # artifacts (neutral accent)
EDGE      = "#C9C7BE"                        # soft box edge
ARROW     = "#333331"
TXT       = "#141413"                        # near-black text for readability
SUB       = "#1A1A19"

plt.rcParams.update({
    # Serif to match the Electronics (Times/Palatino) body text.
    # 'Times New Roman' resolves on Windows; the rest are fallbacks for other machines.
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "Nimbus Roman",
                   "Liberation Serif", "STIXGeneral", "DejaVu Serif"],
    "mathtext.fontset": "stix",     # serif math so $I$, $P$, $V$, $\Pi$ match the body
    "font.size": 11,
    "svg.fonttype": "none",
})
```

```python
fig, ax = plt.subplots(figsize=(11.5, 5.4))
ax.set_xlim(0, 23)
ax.set_ylim(0, 10)
ax.axis("off")

def tile(ax, cx, cy, w, h, fc, bar, title, sub=None):
    """Rounded tile with a left accent bar; bold title, optional subtitle."""
    ax.add_patch(FancyBboxPatch(
        (cx - w/2, cy - h/2), w, h,
        boxstyle="round,pad=0.02,rounding_size=0.14",
        fc=fc, ec=EDGE, lw=1.3, mutation_aspect=1, zorder=2))
    # left accent bar (rounded look via thin rect inset)
    ax.add_patch(Rectangle((cx - w/2 + 0.06, cy - h/2 + 0.12), 0.16, h - 0.24,
                           fc=bar, ec="none", zorder=3))
    if sub:
        ax.text(cx + 0.15, cy + 0.20, title, ha="center", va="center",
                color=TXT, fontsize=12.5, fontweight="bold", zorder=4)
        ax.text(cx + 0.15, cy - 0.24, sub, ha="center", va="center",
                color=SUB, fontsize=11.0, fontweight="bold", zorder=4)
    else:
        ax.text(cx + 0.15, cy, title, ha="center", va="center",
                color=TXT, fontsize=12.5, fontweight="bold", zorder=4)

def arrow(ax, p0, p1, rad=0.0):
    ax.add_patch(FancyArrowPatch(
        p0, p1, arrowstyle="-|>", mutation_scale=16,
        lw=1.7, color=ARROW, shrinkA=3, shrinkB=3,
        connectionstyle=f"arc3,rad={rad}", zorder=1))

# ---- coordinates ----
y1, y2 = 7.6, 2.7
xs = [2.6, 6.6, 10.3, 13.8, 17.4, 20.6]

# top row
tile(ax, xs[0], y1+1.05, 3.2, 1.35, INPUT_FC, INPUT_BAR, "EWF source", "image $I$")
tile(ax, xs[0], y1-1.05, 3.2, 1.35, INPUT_FC, INPUT_BAR, "HVEA graph", "$G_{os}$")
tile(ax, xs[1], y1, 3.5, 1.75, PROC_FC, PROC_BAR, "Phase 1 · Reasoner", "fiwalk + HVEA")
tile(ax, xs[2], y1, 2.7, 1.5, DATA_FC, DATA_BAR, "Reasoning", "trace $R$")
tile(ax, xs[3], y1, 3.0, 1.75, PROC_FC, PROC_BAR, "Phase 2", "Profile compiler")
tile(ax, xs[4], y1, 2.9, 1.5, DATA_FC, DATA_BAR, "SIFTING", "profile $\\Pi$")

# bottom row
tile(ax, xs[1], y2, 2.7, 1.5, DATA_FC, DATA_BAR, "8-metric", "report $V$")
tile(ax, xs[2], y2, 3.0, 1.75, PROC_FC, PROC_BAR, "Phase 4", "Validator")
tile(ax, xs[3], y2, 2.9, 1.5, DATA_FC, DATA_BAR, "Partial", "image $P$")
tile(ax, xs[4], y2, 3.0, 1.75, PROC_FC, PROC_BAR, "Phase 3", "Acquisition harness")

# negative-space audit: DIRECTLY BELOW the reasoning trace (xs[2])
ny = (y1 + y2)/2 + 0.05
tile(ax, xs[2], ny, 2.9, 1.35, DATA_FC, DATA_BAR, "Negative-space", "audit")

# ---- arrows ----
arrow(ax, (xs[0]+1.6, y1+1.05), (xs[1]-1.78, y1+0.35))
arrow(ax, (xs[0]+1.6, y1-1.05), (xs[1]-1.78, y1-0.35))
arrow(ax, (xs[1]+1.78, y1), (xs[2]-1.38, y1))
arrow(ax, (xs[2]+1.38, y1), (xs[3]-1.52, y1))
arrow(ax, (xs[3]+1.52, y1), (xs[4]-1.48, y1))
arrow(ax, (xs[4], y1-0.78), (xs[4], y2+0.9))                 # wrap
arrow(ax, (xs[4]-1.52, y2), (xs[3]+1.48, y2))
arrow(ax, (xs[3]-1.48, y2), (xs[2]+1.52, y2))
arrow(ax, (xs[2]-1.52, y2), (xs[1]+1.38, y2))
arrow(ax, (xs[2], y1-0.78), (xs[2], ny+0.72))               # trace -> neg-space (straight down)

plt.tight_layout()
import os
os.makedirs(OUT_DIR, exist_ok=True)
fig.savefig(os.path.join(OUT_DIR, "hvea_architecture.pdf"), bbox_inches="tight")
fig.savefig(os.path.join(OUT_DIR, "hvea_architecture.png"), dpi=300, bbox_inches="tight")
plt.show()
print("saved to", OUT_DIR)
```
